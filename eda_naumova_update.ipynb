6.**Постройте матрицу корреляций числовых переменных.**

```python
#все числовые переменные
numeric_columns = [
    'station_load_norm', 'n_taxi_start', 'n_taxi_end', 'n_carsharing_start', 'n_carsharing_end',
    'number_of_stations'
]
```

Для построения использую [хитмапу](https://seaborn.pydata.org/generated/seaborn.heatmap.html) из библиотеки sns

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14, 10))
sns.heatmap(data[numeric_columns].corr(), annot=True, fmt=".2f", vmin=-1, vmax=1)
plt.title('Матрица корреляций числовых переменных')
plt.show()
```

Исходя из этого графика можно заметить следующие зависимости:

1) Количесто станций метро имеет обратную зависимость от загруженности метро, те чем меньше количество станций в районе, тем больше загруженность метро

2) Также можно увидеть, что корелляция между n_taxi_end, n_taxi_start, n_carsharing_start и n_carsharing_end - сильная, положительная корелляция, это может означать, что в районах, в которых большой спрос на такси, также большой спрос и на каршеринг(районы в которых высокий спрос на такси/каршеринг в начале/конце, также имеют высокий спрос на такси/каршеринг в конце/начале)

3)Также есть зависимость корелляция между количеством станций метро в районе и количеством поездок на такси, можно сделать вывод, что в районах в которых несколько станций метро больше спрос на такси(начало поездок/конец)

Возможно можно было бы посмотреть на то как зависит загруженность метро от количества станций метро

```python
dta = data.copy()
```

```python
plt.figure(figsize=(10, 5))
sns.scatterplot(x='number_of_stations', y='station_load', data=dta, c='green')
plt.title('Зависимость загруженности от количества станций')
plt.xlabel('Количество станций')
plt.ylabel('Загруженность')
plt.show()
```

Тут убеждаемся, что чем меньше количество станций, тем больше загруженность(зависит также от часа пик)

Также интересно было бы глянуть на среднюю загруженность станций по месяцам(какие месяцы имеют самую высокую загруженность)

```python
dta['month'] = dta['timestamp'].dt.month
month_load = dta.groupby('month')['station_load'].mean()

plt.figure(figsize=(10, 5))
month_load.plot(kind='line', title='Средняя загруженность станций по месяцам')
plt.xlabel('Месяц')
plt.ylabel('Загруженность')
plt.show()
```

Мы видим, что средняя загруженность летом и зимой достигает пика, что вполне оправданно учитывая пробки в данные сезоны и что спрос на метро повышается

Также имеет смысл задаться вопросом о том, как загруженность метро влияет на количество поездок

```python
grouped = dta.groupby('station_load')[['n_taxi_start', 'n_taxi_end', 'n_carsharing_start', 'n_carsharing_end']].mean().reset_index()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='station_load', y='n_taxi_start', data=grouped, hue='station_load', palette='pastel', legend=False)
plt.title('n_taxi_start по уровню загруженности метро')
plt.xlabel('Уровень загруженности метро')
plt.ylabel('n_taxi_start')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='station_load', y='n_taxi_end', data=grouped, hue='station_load', palette='pastel', legend=False)
plt.title('n_taxi_end по уровню загруженности метро')
plt.xlabel('Уровень загруженности метро')
plt.ylabel('n_taxi_end')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='station_load', y='n_carsharing_start', hue='station_load', data=grouped, palette='pastel', legend=False)
plt.title('n_carsharing_start по уровню загруженности метро')
plt.xlabel('Уровень загруженности метро')
plt.ylabel('n_carsharing_start')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.barplot(x='station_load', y='n_carsharing_end', hue='station_load', data=grouped, palette='pastel', legend=False)
plt.title('n_carsharing_end по уровню загруженности метро')
plt.xlabel('Уровень загруженности метро')
plt.ylabel('n_carsharing_end')
plt.show()
```

Ну судя по графику, при маленькой загруженности метро, самый высокий спрос как на такси, так и на каршеринг, при средней загруженности, спрос на такси/каршеринг уменьшается(тут может быть много факторов от пробок до погоды), но что точно можно заметить - при повышении загруженности метро - спрос на такси/каршеринг растет линейно
