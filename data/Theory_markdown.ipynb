### Факторы изменения спроса на такси 
### <small>1. Погодные условия</small>
Неблагоприятные погодные условия увеличивают спрос на такси из-за сокращения использования общественного транспорта. Так, например, исследование данных Uber в Нью-Йорке показало рост количества поездок на такси во время дождя на 22%. Такие выводы подчеркивают важность интеграции метеорологических данных в модели прогнозирования.

Brodeur A., Nield K.  Journal of Economic Behavior & Organization, 152:1–16, 2018. - **"An empirical analysis of taxi, Lyft and Uber rides: Evidence from weather shocks in NYC."**   
> *"Using all Uber and Lyft rides in NYC, we investigated the effect of rain on Uber and Lyft rides and found that the number of Uber (Lyft) rides per hour increases by approximately 22 (19)% when it is raining."*

D. Chen, Y. Zhang, L. Gao, N. Geng, and X. Li.PLoS ONE, 12(9), 2017. – **"The impact of rainfall on the temporal and spatial distribution of taxi passengers"**  
> *"As rainfall intensity increases, the taxi service demands for evening rush hour and non-rush hour periods on workdays show opposite trends."*

Singhal, A., Kamga, C. & Yazici, A.Transportation Research Part A: Policy and Practice, Vol. 69, 379–391, 2014. – **"Impact of weather on urban transit ridership"** 
> *"The paper endorses previous studies in recognizing the adverse impact of weather conditions like rain and snow on transit ridership and the finding that ridership on weekends is more severely affected."*

### <small>2. Загруженность и доступность метро</small>
Спрос на такси тесно связан с доступностью метро: в зонах с хорошо развитой транспортной инфраструктурой реже выбирают такси, тогда как в районах с ограниченным доступом к метро оно становится более востребованным, поэтому при анализе этот фактор будет учтен.

M. Li, L. Dong, Z. Shen, W. Lang, and X. Ye. Sustainability, vol. 9, no. 2, pp. 1–12, 2017.- **"Examining the Interaction of Taxi and Subway Ridership for Sustainable Urbanization."**
> *".. increases in travel time and distance would increase the demand for subway, while taxi trips largely represented movements for those locations that the subway could not reach …"*

S. Jiang, G. Wei, Z. He, and L. Yang. Journal of Advanced Transportation, 2018(2063):1–14, 2018. - **"Exploring the intermodal relationship between taxi and subway in Beijing, China."**  
> *"The results of different negative binomial regression models for different situations indicate that subway ridership plays an important role in explaining differences in taxi trips"*

C. Chen, T. Feng, C. Ding, B. Yu, and B. Yao. . Journal of Transport Geography, 96: Article 103172, 2021. - **"Examining the spatial-temporal relationship between urban built environment and taxi ridership: Results of a semi-parametric GWPR model."** 
> *"... the well-connected public transport system, reduce the reliance on taxis, making other transportation options, such as the subway and buses, more attractive for commuters ..."*

### <small>3. Загруженность дорог</small>
Издержки поездок на общественном или собственном транспорте в условиях повышенной загруженности (потеря времени в пробках, проблемы с поиском парковочного места, а также стресс от вождения в городских условиях) приводят к сдвигу потребительских предпочтений в сторону более удобной альтернативы - такси. Использование данных о пробках позволит точнее спрогнозировать рост спроса и оптимизировать распределение машин.

**Liu, Xinmin & Sun, Lu & Sun, Qiuxia & Gao, Ge.Journal of Advanced Transportation. 1-20, 2020. - **"Spatial Variation of Taxi Demand Using GPS Trajectories and POI Data."**  
> *"... impact of road density on taxi demand varies greatly on weekdays, with little difference between weekends. Places with higher road density may generate more taxi trips, and these places usually have a higher population density"*

K.I. Wong, S.C. Wong, Hai Yang, J.H. Wu. Transportation Research Part B: Methodological.985-1000,2008. - **"Modeling urban taxi services with multiple user classes and vehicle modes."**
> *"Although the utilization of normal taxis increases at a lower rate than the scaling factor, it does not decrease (and is expected to approach the maximum utilization rate, or the total taxi service time), because the increasing effect on the total trip time due to increased total demand and traffic congestion should always dominate the decreasing effect because of decreased demand through traffic congestion and thus a longer customer waiting time."*

Zhang, Man & Tian, Dongwei & Liu, Jingming & Li, Xuehua. Sustainability. 16,(2024). - **"Analysis of Taxi Demand and Traffic Influencing Factors in Urban Core Area Based on Data Field Theory and GWR Model: A Case Study of Beijing."**  
> *"Commuters often favor taxis, as they provide a quicker way to commute to work, avoiding the traffic congestion and parking issues associated with driving"*

### <small>4. Время поездки (время суток, день недели, праздничные дни)</small>
Время поездки играет ключевую роль в прогнозировании спроса на такси, поскольку разные временные интервалы связаны с различными типами активности людей.

Yang, C., & Gonzales, E. J.Transportation Research Record, 2429, 203-212,2014. - **"Modeling Taxi Trip Demand by Time of Day in New York City."**  
> *"... the total number of jobs has a higher influence on taxi demand from 7 a.m. to 6 p.m., which indicates that extra taxi demand during this period in NYC is likely caused by people going to and from work or work-related activities."*

W. Chen, Y. Li, and Q. Zhang. Journal of Advanced Transportation, vol. 2020, pp. 1–13, Jul. 2020. - **"Taxi demand prediction based on a combination forecasting model in hotspots."** 
> *"the most important influencing factor of the taxi demand prediction model is the time factor"*

Naji, Hasan A. H., Xue, Qingji, Zhu, Huijun, Li, Tianfeng. Applied Sciences, 11 (20), Oct.2021. - **"Forecasting Taxi Demands Using Generative Adversarial Networks with Multi-Source Data."** 
> *"We can see that these areas have similar taxi demand distributions (especially on Fridays, indicating that people take taxis on the last weekday for reasons such as going to travel or for entertainment or coming back on Sundays) and similar changes in taxi demands. Therefore, we can leverage the weekdays and weekends data to help in forecasting taxi demands."*

### <small>5. Социально-экономический фактор</small>
Приведенные исследования подтверждают наличие корреляции между спросом на такси и социально-экономическими показателями: уровнем жизни, плотностью населения, уровнем безработицы и другими. Использование таких показателей вместе с геопространственными данными позволит повысить точность прогнозирования спроса.

Liu, Xinmin & Sun, Lu & Sun, Qiuxia & Gao, Ge. Journal of Advanced Transportation.1-20, 2020. - **"Spatial Variation of Taxi Demand Using GPS Trajectories and POI Data."**  
> *"residential density, housing price and road density are positively correlated with higher demand for taxi, but land use mix, residential area shows negatively on travel demand"*

Yan, J., Xiang, L., Wu, C., and Wu, H. Int. Arch. Photogramm. Remote Sens. Spatial Inf. Sci., XLIII-B4, 213–220,2020. - **"CITY-SCALE TAXI DEMAND PREDICTION USING MULTISOURCE URBAN GEOSPATIAL DATA."** 
> *"Table 1 shows that the drop-off amount and POI amount are both strongly related to taxi demand, while road and population are moderately related. Thus, combining the geospatial data properly can assist taxi demand prediction to a certain degree."*

Gangrade, Akshata & Pratyush, Pawel & Hajela, Gaurav.Etri Journal, 2022. - **"Taxi‐demand forecasting using dynamic spatiotemporal analysis."**   
> *"taxi demand can be expected to be greater in areas where per-capita income is higher and where other factors—such as percent households below poverty, percent aged 16+ unemployed, hardship index, and others—are lower, and vice versa"*
