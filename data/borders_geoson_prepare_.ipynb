```python
import folium
import json
import pandas as pd
from folium import Choropleth
# Загружаем GeoJSON с границами районов
with open('export_borders.geojson', 'r', encoding='utf-8') as f:
    geojson_data = json.load(f)

# Фильтруем районы, у которых есть название (исключаем "Нет названия")
filtered_features = [
    feature for feature in geojson_data['features']
    if 'name' in feature['properties'] and feature['properties']['name'] not in [None, 'Нет названия']
]

# Создаем новый GeoJSON с отфильтрованными районами
msc_borders_clean = {
    "type": "FeatureCollection",
    "features": filtered_features
}
# Сохраняем обновленный файл
#with open('msc_borders_clean.geojson', 'w', encoding='utf-8') as f:
    #json.dump(msc_borders_clean, f, ensure_ascii=False, indent=4)
```
